```python
import sys
from pathlib import Path
import pandas as pd
# Añadir la raíz del proyecto al path
sys.path.append(str(Path.cwd().parent))

from src.storage.minio_client import *
import io
```

```python
# 1. Leer Raw
data = read_json(
    bucket_name="raw",
    object_name="networks.json"
)
```

```python
df = pd.DataFrame(data["networks"])
```

```python
# 2. Convertir .json → Parquet
buffer = io.BytesIO()

df.to_parquet(
    buffer,
    index=False
)

buffer.seek(0)
```

```python
# 3. Guardar en Bronze
upload_bytes(
    bucket_name="bronze",
    object_name="networks.parquet",
    data=buffer.getvalue(),
    content_type="application/octet-stream"
)
```
