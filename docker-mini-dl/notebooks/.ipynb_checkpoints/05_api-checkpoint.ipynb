```python
import sys
from pathlib import Path

# Añadir la raíz del proyecto al path
sys.path.append(str(Path.cwd().parent))

from src.storage.minio_client import *
import io
```

```python
# 1. Leer Raw
df = read_csv(
    bucket_name="raw",
    object_name="ventas.csv"
)
```

```python
# 2. Convertir CSV → Parquet
buffer = io.BytesIO()

df.to_parquet(
    buffer,
    index=False
)

buffer.seek(0)
```

```python
print(f"Tamaño del Parquet: {buffer.getbuffer().nbytes / 1024:.2f} KB")
```

```python
# 3. Guardar en Bronze
upload_bytes(
    bucket_name="bronze",
    object_name="ventas.parquet",
    data=buffer.getvalue(),
    content_type="application/octet-stream"
)
```
